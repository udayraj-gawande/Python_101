Discount and Payble Amount .....


```python
price = float(input("Enter products prices:-"))

if (price>0):
    if price>=5000:
        Amt = price*0.15
        Payble = price - Amt
        print("Discount Amt:-",Amt)
        print("Payble Amt",Payble)
    elif price>=1000:
        Amt = price*0.10
        Payble = price - Amt
        print("Discount Amt:-",Amt)
        print("Payble Amt",Payble)
    elif price>=500:
        Amt = price*0.06
        Payble = price - Amt
        print("Discount Amt:-",Amt)
        print("Payble Amt",Payble)
    else:
        print("Not Valid")
else :
    print("Not Valid price")
```
